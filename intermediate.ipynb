Exercises
1. Extract email domains and find the most common ones
2. Count how many phone numbers in Phone 1 column start with +1 (US format)
3. Group subscriptions by month and year to see trends
4. Find last names that appear more than once
5. Extract top-level domains from websites and count them

```python
import pandas as pd

path = input("CSV path: ")
df = pd.read_csv(path)
domains = df['Email'].str.split('@').str[1]
domain_counts = domains.value_counts().head(10)
print(domain_counts)
```

```python
us_format_phones = df['Phone 1'].str.startswith('+1').sum()
print(f"Phone numbers in US format: {us_format_phones}")
```

```python
df['Subscription Date'] = pd.to_datetime(df['Subscription Date'])
by_year = df['Subscription Date'].dt.year.value_counts()
by_month = df['Subscription Date'].dt.month.value_counts()
print(by_year)
print(by_month)
```

```python
lastnames = df['Last Name'].value_counts()
duplicate_lastnames = lastnames[lastnames > 1]
print(duplicate_lastnames)
```

```python
tld = df['Email'].str.split('.').str[1]
print(tld.value_counts())
```
