Exercises
1. Check if all Customer IDs have the same length and are alphanumeric
2. Find which country has the most unique companies
3. Find the most common first names and last names
4. Find the earliest and latest subscription dates
5. Find customers from specific countries who subscribed in 2022 and have a .com website
6. Check for any missing values

```python
import pandas as pd

df = pd.read_csv('customers-100.csv')
df['ID Length'] = df['Customer Id'].str.len()
df['Is Alphanumeric'] = df['Customer Id'].str.isalnum()

print(f"Unique ID lengths: {df['ID Length'].unique()}")
print(f"Non-alphanumeric IDs: {df[~df['Is Alphanumeric']]['Customer Id'].tolist()}")
```

```python
unique_companies_by_country = df.groupby('Country')['Company'].nunique().sort_values(ascending=False).reset_index(name='Unique Companies Count')
display(unique_companies_by_country.head(10))
```

```python
print(df['First Name'].value_counts().head())
print(df['Last Name'].value_counts().head())
```

```python
df['Subscription Date'] = pd.to_datetime(df['Subscription Date'])
earliest_date = df['Subscription Date'].min()
latest_date = df['Subscription Date'].max()

print(f"Earliest Subscription Date: {earliest_date.strftime('%Y-%m-%d')}")
print(f"Latest Subscription Date: {latest_date.strftime('%Y-%m-%d')}")
```

```python
filtered_customers = df[
    (df['Subscription Date'].dt.year == 2022) &
    (df['Website'].str.endswith('.com') | df['Website'].str.endswith('.com/'))
]

print(filtered_customers[['First Name', 'Last Name', 'Country', 'Subscription Date', 'Website']].sort_values(by='Country'))
```
